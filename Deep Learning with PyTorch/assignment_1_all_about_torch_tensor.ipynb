```python
# Import torch and other required modules
import torch
```

## Function 1 - torch.squeeze

Returns a tensor with all specified dimensions of input of size 1 removed.

```python
x = torch.zeros(5, 3, 1, 3, 2, 1)
print(x.shape)
print(x.squeeze().shape)
```

Removes all brackets of size 1

```python
# Example 2 - working
x = torch.zeros(5, 3, 1, 3, 2, 1)
print(x.shape)
print(x.squeeze([2, 3]).shape)
```

Removes specified index of dimensions if their size is 1

```python
# Example 3 - breaking (to illustrate when it breaks)
x = torch.zeros(5, 3, 1, 3, 2, 2)
print(x.shape)
print(x.squeeze(dim=7).shape)
```

Dimension to remove is out of range

Use the function when need to reduce the matrix sizes by removing unnecessary brackets

## Function 2 - torch.unsqueeze()

Add dimension of size 1 to the specific position

```python
x2 = torch.tensor([[1, 2, 3, 4], [5, 6, 7, 8]])
print(x2)
print(x2.size())
print("="*30)

x2 = x2.unsqueeze(1)
print(x2)
print(x2.size())
print("="*30)

x2 = x2.squeeze().unsqueeze(2)
print(x2)
print(x2.size())
```

Inserts brackets to a specific index in the tensor

```python
# Example 2 - working
x2 = torch.tensor([[1, 2, 3, 4], [5, 6, 7, 8]])
print(x2)
print(x2.size())
print("="*30)

x2 = x2.unsqueeze(-1)
print(x2)
print(x2.size())
print("="*30)

x2 = x2.squeeze().unsqueeze(-3)
print(x2)
print(x2.size())
```

indexing the unsuqeeze value works the same way as with array indexing

```python
# Example 3 - breaking (to illustrate when it breaks)
x2 = torch.tensor([[1, 2, 3, 4], [5, 6, 7, 8]])
print(x2)
print(x2.size())
print("="*30)

x2 = x2.unsqueeze(dim=[2, 3])
print(x2)
print(x2.size())
print("="*30)
```

Unlike `torch.squeeze()` dim parameter takes only int values

Usefule for operation with tensors which require specific dimension's size

## Function 3 - torch.rand()

Returns a tensor filled with random numbers from a uniform distribution on the interval `[0, 1)`

```python
# Example 1 - working
print(torch.rand(3))
```

Creates a vector of size 3 with random values

```python
# Example 2 - working
x3 = torch.rand(2, 3, 4)
print(x3)
x3.size()
```

Creates a matrix of the size (2, 3, 4)

```python
# Example 3 - breaking (to illustrate when it breaks)
x3 = torch.rand(-1)
x3
```

The size of the desired tensor should be positive

Usefule for fast checking and creating initial values for further algorithm

## Function 4 - torch.cat()

Concatenates the given sequence of tensors in tensors in the given dimension. All tensors must either have the same shape (except in the concatenating dimension) or be a 1-D empty tensor with size (0,).

```python
# Example 1 - working
x4 = torch.randint(10, (2, 3))
y4 = torch.randint(10, (1, 3))
z4 = torch.cat((x4, y4), 0)
print(z4)
print(z4.size())
```

Concatenates rows

```python
# Example 2 - working
x4 = torch.randint(10, (2, 3))
y4 = torch.randint(10, (2, 5))
z4 = torch.cat((x4, y4), 1)
print(z4)
print(z4.size())
```

Concatenates by columns

```python
# Example 3 - breaking (to illustrate when it breaks)
x4 = torch.randint(10, (2, 3, 3))
y4 = torch.randint(10, (2, 3, 4))
z4 = torch.cat((x4, y4), 1)
print(z4)
print(z4.size())
```

To be concatenated tensors should have the same dimension except for one by which the operation is conducted

To combine data points for the specific task (e.g. preprocessing)

## Function 5 - torch.hsplit()

Splits vector/matrix horizontally (same rows diff columns)

```python
# Example 1 - working
x5 = torch.arange(20, dtype=torch.int64).reshape(2, 10)
print(x5)
print("="*50)

x6, x7 = torch.hsplit(x5, 2)
print(x6, x7, sep='\n')
```

Splits into 2 equal tensors horizontally

```python
# Example 2 - working
x5 = torch.arange(20, dtype=torch.int64).reshape(2, 10)
print(x5)
print("="*50)

x51, x52, x53 = torch.hsplit(x5, [3, 10])
print(x51, x52, x53, sep='\n')
```

[3, 10] Means that splitting points are columns' indexes [0, 2], [3, 9], []

Since max idx is 9, then second tensor is [3, 9] and the next one is blank

```python
# Example 3 - breaking (to illustrate when it breaks)
# Example 2 - working
x5 = torch.arange(20, dtype=torch.int64).reshape(2, 10)
print(x5)
print("="*50)

x5 = torch.hsplit(x5, 6)
print(x5, sep='\n')
```

If indices_or_sections is int and its value cannot be divided by the column size, there will be an error

To split features and target etc
